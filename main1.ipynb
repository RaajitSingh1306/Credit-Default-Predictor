```python
df = pd.read_csv('loan_data(best).csv')

print(df.shape)
print(df.dtypes)
print(df.describe)
```

```python
cat_col=['person_gender','person_education',
        'person_home_ownership', 'loan_intent',
        'previous_loan_defaults_on_file']

for col in cat_col:
    print(f"\n{col}")
    print(df[col].value_counts())
```

```python
print("age over 100:", (df['person_age']>100).sum())
print("age over 80 :", (df['person_age']>80).sum())

print("person with exp over 60 :", (df['person_emp_exp']>60).sum())
print("person with exp over 40 :", (df['person_emp_exp']>40).sum())

print("Income over 1M:", (df["person_income"]>1000000).sum())
print("Income over 500K:", (df["person_income"]>500000).sum())

df['age_exp_gap'] = df['person_age']-df['person_emp_exp']
print("Rows where emp_exp >= age:", (df['age_exp_gap'] <= 0).sum())

original_size=len(df)
print(original_size)
```

```python
#fix age and income
df = df[df['person_age']<=100]
df = df[df['person_emp_exp']<=60]

income_cap = df['person_income'].quantile(0.99)
df = df[df['person_income']<=income_cap]

df = df.drop(columns='age_exp_gap')
```

```python
# Verify
print("Rows removed:",original_size-len(df))
print("Rows remaining:",len(df))

print(f"New age range: {df['person_age'].min()} to {df['person_age'].max()}")
print(f"New exp range: {df['person_emp_exp'].min()} to {df['person_emp_exp'].max()}")
print(f"New income range: {df['person_income'].min():,.0f} to {df['person_income'].max():,.0f}")
```
